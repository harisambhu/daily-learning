# create Dataset

```python
import pandas as pd

data = {
    'Age': [22, 24, 26, 28, 30, 32, 35, 38, 40, 42, 45, 48],

    'Income': [30000, 30000, 50000, 50000, 50000, 50000,
               70000, 70000, 70000, 70000, 80000, 80000],

    'Experience': [1, 6, 2, 8, 3, 10,
                   4, 9, 5, 11, 12, 15],

    'Buy': ['No', 'Yes', 'No', 'Yes', 'No', 'Yes',
            'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes']
}

df = pd.DataFrame(data)
df
```

# separate feture and target 

```python
x = df[['Age', 'Income', 'Experience']]
y = df['Buy']
```

# display feature and target 

```python

print("Features (X):")

print("Target (y):")
```

# create the decision tree model 

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=2,      # Limit to 2 levels of decision nodes
    random_state=42
)

```

# Step 6: Train the Model

```python
model.fit(x, y)
```

# predict a new customer 

```python
# New customer data
new_customer = [[35, 60000, 8]]

# Predict
prediction = model.predict(new_customer)

print("Prediction:", prediction)
```

# predict multiple customer 

```python
# Multiple new customers
new_customers = [
    [25, 30000, 2],
    [35, 60000, 8],
    [45, 80000, 15],
    [30, 50000, 5]
]

# Predict
predictions = model.predict(new_customers)

print("Predictions:", predictions)
```

# visualition the decision tree

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
plt.figure(figsize=(10, 6))

plot_tree(
    model,
    feature_names=x.columns,
    class_names=["No", "Yes"],   # Change if your classes are different
    filled=True,
    rounded=True,
    fontsize=10
)

plt.title("Decision Tree")
plt.show()
```
